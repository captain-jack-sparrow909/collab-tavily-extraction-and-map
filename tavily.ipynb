```python
import asyncio
import os
import json
import ssl
from datetime import datetime

import certifi
from langchain_tavily import TavilyExtract, TavilyMap
from rich.console import Console
from rich.table import Table
from rich.panel import Panel
from rich.progress import Progress

ssl_context = ssl.create_default_context(cafile=certifi.where())
os.environ["SSL_CERT_FILE "] = certifi.where()
os.environ[" REQUESTS_CA_BUNDLE"] = certifi.where()

console = Console()

print("All imported successfully")
```

```python
tavily_map = TavilyMap(max_depth=3, max_breadth=15, limit=500)
```

```python
demo_url = "https://python.langchain.com/docs/introduction/"
site_map = tavily_map.invoke(demo_url)
urls = site_map.get("results", [])
print("length of urls", len(urls))
```

```python
print(urls[0])
```

```python
tavily_extract = TavilyExtract()
print("tavily extract ..")
```

```python
sample_urls = [urls[20]]
extraction_result = tavily_extract.ainvoke({"input": sample_urls})
print(extraction_result)
extracted_docs = extraction_result.results

for i, doc in enumerate(extracted_docs, 1):
  url = doc.get("url", 'Unkown')
  content = doc.raw_content
  console.print(Panel(panel_content=f"URL: {url} content length: {len(content)}", title = f"Document {i}", border_style="blue"))
  print()
```
